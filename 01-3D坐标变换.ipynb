# 机器人空间描述与坐标变换

在机器人学中，需要用坐标系描述物体的位置和姿态。本节介绍旋转矩阵、平移向量和齐次变换矩阵，并区分刚体运动与坐标系变换的物理含义。

## 1. 旋转与齐次变换

### 1.1 旋转矩阵

二维或三维空间中的旋转可以通过正交矩阵描述。以绕 $x/y/z$ 轴旋转 $\theta$ 角为例，其旋转矩阵 $R_*(\theta)$ 为：

$$
R_x(\theta) =
\begin{bmatrix}
1 & 0 & 0 \\
0 & \cos\theta & -\sin\theta \\
0 & \sin\theta & \cos\theta
\end{bmatrix}
$$

$$
R_y(\theta) =
\begin{bmatrix}
\cos\theta & 0 & \sin\theta \\
0 & 1 & 0 \\
-\sin\theta & 0 & \cos\theta
\end{bmatrix}
$$

$$
R_z(\theta) =
\begin{bmatrix}
\cos\theta & -\sin\theta & 0 \\
\sin\theta & \cos\theta & 0 \\
0 & 0 & 1
\end{bmatrix}
$$


## 2. 齐次变换矩阵

如果空间是 $n$ 维，那么：

- 点：$n$ 维向量 $p \in \mathbb{R}^n$
- 线性变换（旋转、缩放等）：$n\times n$ 矩阵 $A$
- 平移：$n\times1$ 向量 $t$
- 齐次坐标：$(n+1)$ 维向量 $\tilde{p} = \begin{bmatrix} p \\ 1 \end{bmatrix}$
- 齐次变换矩阵：$(n+1)\times(n+1)$

结构为：

$$
T =
\begin{bmatrix}
A & t \\
0 & 1
\end{bmatrix}
$$

其中 $0$ 是 $1\times n$ 的零行。

为了将旋转 ($R, 3\times3$) 和平移 ($t, 3\times1$) 统一在线性运算中，引入了 $4\times4$ 的齐次变换矩阵 $T$：

$$
T =
\begin{bmatrix}
R & t \\
0 & 1
\end{bmatrix}
=
\begin{bmatrix}
r_{11} & r_{12} & r_{13} & t_x \\
r_{21} & r_{22} & r_{23} & t_y \\
r_{31} & r_{32} & r_{33} & t_z \\
0 & 0 & 0 & 1
\end{bmatrix}
$$

其中 $t_x, t_y, t_z$ 是平移向量 $t$ 的分量。


## 3. 齐次变换公式

$$
\tilde{p}' = T \tilde{p}
$$

其中：

- $\tilde{p}$：变换前的齐次坐标
- $T$：齐次变换矩阵
- $\tilde{p}'$：变换后的齐次坐标

#### 展开形式

一般 $n$ 维空间中：

$$
\tilde{p} = \begin{bmatrix} p \\ 1 \end{bmatrix},
\quad
T = \begin{bmatrix} A & t \\ 0 & 1 \end{bmatrix}
$$

代入公式：

$$
\tilde{p}' = T \tilde{p}
=
\begin{bmatrix} A & t \\ 0 & 1 \end{bmatrix}
\begin{bmatrix} p \\ 1 \end{bmatrix}
=
\begin{bmatrix} A p + t \\ 1 \end{bmatrix}
$$

去掉最后的 $1$，得到普通坐标下的结果：

$$
p' = A p + t
$$

三维刚体变换中，$A = R$ 是旋转矩阵：

$$
p' = R p + t
$$

含义：

- 先对点 $p$ 做线性变换 $A$（通常是旋转）
- 再平移 $t$
- 整个过程通过一次矩阵乘法 $T \tilde{p}$ 完成


## 4. 主动变换与坐标变换

齐次变换矩阵

$$
T =
\begin{bmatrix}
R & t \\
0 & 1
\end{bmatrix}
$$

在数学形式上表示旋转和平移的组合，但在机器人学中，需要特别区分它的两种物理含义：

1. **主动变换（Active Transformation）**：描述物体或点在空间中真的发生了运动。
2. **坐标变换（Coordinate Transformation / Passive Transformation）**：描述同一个物理点在不同坐标系下的坐标表示。

虽然两种情况都可能使用齐次变换矩阵，但它们的物理含义不同。

### 4.1 主动变换：物体真的发生运动

假设坐标系保持不变，点 $P$ 在空间中经过旋转和平移，从原来的位置运动到新的位置。

可以写成：

$$
\tilde{p}' = T\tilde{p}
$$

展开：

$$
p' = Rp+t
$$

其中：

- $p$：点运动前的坐标
- $p'$：点运动后的坐标
- $R$：对点进行的旋转
- $t$：对点进行的平移

此时 $p$ 和 $p'$ 对应的是**同一个坐标系中，物理点运动前后的两个不同位置**。

即：

$$
\boxed{\text{坐标系不动，物体/点发生运动}}
$$

---

### 4.2 坐标变换：物理点不动，只改变坐标表示

在机器人学中，更常见的情况是存在多个坐标系。

假设：

- $\{A\}$：基坐标系
- $\{B\}$：机器人末端坐标系
- $P$：空间中的某个物理点

点 $P$ 本身没有发生任何运动，但是在不同坐标系下，它的坐标数值不同。

记：

$$
P_B
$$

为点 $P$ 在坐标系 $\{B\}$ 下的坐标，

$$
P_A
$$

为同一个点 $P$ 在坐标系 $\{A\}$ 下的坐标。

如果 $T_{AB}$ 描述坐标系 $\{B\}$ 相对于坐标系 $\{A\}$ 的位姿，则：

$$
\boxed{P_A=T_{AB}P_B}
$$

展开后：

$$
P_A=R_{AB}P_B+t_{AB}
$$

其中：

- $R_{AB}$：坐标系 $\{B\}$ 相对于 $\{A\}$ 的朝向
- $t_{AB}$：坐标系 $\{B\}$ 原点在 $\{A\}$ 中的位置
- $P_B$：物理点 $P$ 在 $\{B\}$ 中的坐标表示
- $P_A$：同一个物理点 $P$ 在 $\{A\}$ 中的坐标表示

这里并不是把物理点 $P$ 从一个位置移动到了另一个位置，而是：

$$
\boxed{\text{物理点不动，只改变描述这个点所使用的坐标系}}
$$

也就是：

$$
P_B
\quad\xrightarrow{T_{AB}}\quad
P_A
$$

表示：

$$
\boxed{\text{点 }P\text{ 的 B 系坐标表示}
\rightarrow
\text{同一个点 }P\text{ 的 A 系坐标表示}}
$$

---

### 4.3 两种含义的对比

| 情况 | 坐标系 | 物理点 | 公式含义 |
|---|---|---|---|
| 主动变换 | 不变 | 真的运动 | $p' = Tp$ |
| 坐标变换 | 使用不同坐标系描述 | 不动 | $P_A=T_{AB}P_B$ |

因此需要注意：

> **齐次变换矩阵本身只提供旋转和平移的数学关系。究竟表示“物体真的运动”，还是“同一个物体在不同坐标系下的坐标转换”，取决于具体的物理语境。**

在机器人学中，尤其是机械臂运动学、基坐标系、关节坐标系、末端坐标系之间的关系中，经常使用第二种解释：

$$
\boxed{\text{同一个物理对象不变，在不同坐标系之间转换其坐标表示}}
$$

#### 记忆方式

可以简单记为：

**主动变换：**

$$
\boxed{\text{坐标系不动，点动}}
$$

**坐标变换：**

$$
\boxed{\text{点不动，换坐标系描述}}
$$

二者可能具有相似的矩阵计算形式，但**物理含义不同，不能混为一谈**。

#### 三维具体写法

$$
\tilde{p} =
\begin{bmatrix}
x \\ y \\ z \\ 1
\end{bmatrix},
\quad
T =
\begin{bmatrix}
r_{11} & r_{12} & r_{13} & t_x \\
r_{21} & r_{22} & r_{23} & t_y \\
r_{31} & r_{32} & r_{33} & t_z \\
0 & 0 & 0 & 1
\end{bmatrix}
$$

则：

$$
\tilde{p}' = T \tilde{p}
=
\begin{bmatrix}
r_{11}x + r_{12}y + r_{13}z + t_x \\
r_{21}x + r_{22}y + r_{23}z + t_y \\
r_{31}x + r_{32}y + r_{33}z + t_z \\
1
\end{bmatrix}
$$

取前三个分量，就是变换后的三维点：

$$
p' =
\begin{bmatrix}
x' \\ y' \\ z'
\end{bmatrix}
=
R p + t
$$

#### 记忆要点

1. 点要先写成齐次坐标：

   $$
   p \rightarrow \tilde{p} = \begin{bmatrix} p \\ 1 \end{bmatrix}
   $$

2. 用齐次变换矩阵左乘：

   $$
   \tilde{p}' = T \tilde{p}
   $$

3. 变换后仍然是齐次坐标：

   $$
   \tilde{p}' = \begin{bmatrix} p' \\ 1 \end{bmatrix}
   $$

4. 去掉最后的 $1$，得到普通三维点 $p'$。

5. 矩阵乘法顺序不能反：

   $$
   T \tilde{p} \neq \tilde{p} T
   $$

6. 一句话总结：

   **齐次变换矩阵将旋转和平移统一为一次矩阵乘法；根据具体语境，它既可以描述刚体的实际运动，也可以描述同一物理对象在不同坐标系之间的坐标转换。**


```python
import numpy as np

def rot_x(angle):
    """生成绕X轴的旋转矩阵"""
    theta = np.radians(angle)
    c, s = np.cos(theta), np.sin(theta)
    return np.array([
        [1, 0,  0],
        [0, c, -s],
        [0, s,  c]
    ])

def rot_y(angle):
    """生成绕Y轴的旋转矩阵"""
    theta = np.radians(angle)
    c, s = np.cos(theta), np.sin(theta)
    return np.array([
        [ c, 0, s],
        [ 0, 1, 0],
        [-s, 0, c]
    ])

def rot_z(angle):
    """生成绕Z轴的旋转矩阵"""
    theta = np.radians(angle)
    c, s = np.cos(theta), np.sin(theta)
    return np.array([
        [c, -s, 0],
        [s,  c, 0],
        [0,  0, 1]
    ])

def get_homogeneous_matrix(R, P):
    """组合旋转和平移生成4x4齐次变换矩阵"""
    T = np.eye(4)
    # T[0:3, 0:3]，取0、1、2行和0、1、2列，赋值为旋转矩阵R
    T[:3, :3] = R
    # T[0:3, 3]，取0、1、2行和第3列，赋值为平移向量P
    T[:3, 3] = P
    return T
```

```python
# --- 场景模拟 ---
# 机器人末端坐标系 {B} 相对于基坐标系 {A}:
# 1. 绕 Z 轴旋转 100 度
# 2. 沿 X 轴平移 1 米, 沿 Y 轴平移 2 米

# 1. 定义旋转和平移
R_AB = rot_x(120)
P_AB = np.array([3, 2, 1])

# 2. 构建变换矩阵 T_AB
T_AB = get_homogeneous_matrix(R_AB, P_AB)

# 3. 定义末端上的一个点 P
P_B = np.array([1, 0, 0, 1]) # 齐次坐标

# 4. 计算该点在基坐标系中的位置
P_A = T_AB @ P_B

print("变换矩阵 T_AB:\n", T_AB)
print("-" * 30)
print(f"点在局部坐标系 B: {P_B[:3]}")
print(f"点在全局坐标系 A: {P_A[:3]}")
```
