# 机器人空间描述与坐标变换

在机器人学中，需要用坐标系描述物体的位置和姿态。本节介绍旋转矩阵、平移向量和齐次变换矩阵，并区分刚体运动与坐标系变换的物理含义。

## 1. 旋转与齐次变换

### 1.1 旋转矩阵

二维或三维空间中的旋转可以通过正交矩阵描述。以绕 $x/y/z$ 轴旋转 $\theta$ 角为例，其旋转矩阵 $R_*(\theta)$ 为：

$$
R_x(\theta) =
\begin{bmatrix}
1 & 0 & 0 \\
0 & \cos\theta & -\sin\theta \\
0 & \sin\theta & \cos\theta
\end{bmatrix}
$$

$$
R_y(\theta) =
\begin{bmatrix}
\cos\theta & 0 & \sin\theta \\
0 & 1 & 0 \\
-\sin\theta & 0 & \cos\theta
\end{bmatrix}
$$

$$
R_z(\theta) =
\begin{bmatrix}
\cos\theta & -\sin\theta & 0 \\
\sin\theta & \cos\theta & 0 \\
0 & 0 & 1
\end{bmatrix}
$$


## 2. 齐次变换矩阵

在 $n$ 维空间中，各对象可表示为：

- 点：$n$ 维向量 $p \in \mathbb{R}^n$
- 线性变换（旋转、缩放等）：$n\times n$ 矩阵 $A$
- 平移：$n\times1$ 向量 $t$
- 齐次坐标：$(n+1)$ 维向量 $\tilde{p} = \begin{bmatrix} p \\ 1 \end{bmatrix}$
- 齐次变换矩阵：$(n+1)\times(n+1)$

结构为：

$$
T =
\begin{bmatrix}
A & t \\
0 & 1
\end{bmatrix}
$$

其中 $0$ 是 $1\times n$ 的零行。

旋转矩阵 $R$ 和平移向量 $t$ 可组合为 $4\times4$ 齐次变换矩阵 $T$：

$$
T =
\begin{bmatrix}
R & t \\
0 & 1
\end{bmatrix}
=
\begin{bmatrix}
r_{11} & r_{12} & r_{13} & t_x \\
r_{21} & r_{22} & r_{23} & t_y \\
r_{31} & r_{32} & r_{33} & t_z \\
0 & 0 & 0 & 1
\end{bmatrix}
$$

其中 $t_x, t_y, t_z$ 是平移向量 $t$ 的分量。


## 3. 齐次变换公式

$$
\tilde{p}' = T \tilde{p}
$$

其中：

- $\tilde{p}$：变换前的齐次坐标
- $T$：齐次变换矩阵
- $\tilde{p}'$：变换后的齐次坐标

#### 展开形式

一般 $n$ 维空间中：

$$
\tilde{p} = \begin{bmatrix} p \\ 1 \end{bmatrix},
\quad
T = \begin{bmatrix} A & t \\ 0 & 1 \end{bmatrix}
$$

代入公式：

$$
\tilde{p}' = T \tilde{p}
=
\begin{bmatrix} A & t \\ 0 & 1 \end{bmatrix}
\begin{bmatrix} p \\ 1 \end{bmatrix}
=
\begin{bmatrix} A p + t \\ 1 \end{bmatrix}
$$

去掉最后的 $1$，得到普通坐标下的结果：

$$
p' = A p + t
$$

三维刚体变换中，$A = R$ 是旋转矩阵：

$$
p' = R p + t
$$

运算顺序：

- 对点 $p$ 应用线性变换 $A$（通常为旋转）
- 随后加上平移向量 $t$
- 该过程可通过一次矩阵乘法 $T \tilde{p}$ 表示


## 4. 主动变换与坐标变换

齐次变换矩阵

$$
T =
\begin{bmatrix}
R & t \\
0 & 1
\end{bmatrix}
$$

在数学形式上表示旋转和平移的组合，但在机器人学中，需要区分它的两种物理含义：

1. **主动变换（Active Transformation）**：描述物体或点在空间中发生实际运动。
2. **坐标变换（Coordinate Transformation / Passive Transformation）**：描述同一个物理点在不同坐标系下的坐标表示。

虽然两种情况都可能使用齐次变换矩阵，但它们的物理含义不同。

### 4.1 主动变换：物体发生实际运动

假设坐标系保持不变，点 $P$ 在空间中经过旋转和平移，从初始位置运动到新位置。

可以写成：

$$
\tilde{p}' = T\tilde{p}
$$

展开：

$$
p' = Rp+t
$$

其中：

- $p$：点运动前的坐标
- $p'$：点运动后的坐标
- $R$：对点进行的旋转
- $t$：对点进行的平移

此时 $p$ 和 $p'$ 对应的是**同一个坐标系中，物理点运动前后的两个不同位置**。

即：

$$
\boxed{\text{坐标系保持不变，物体或点发生运动}}
$$

---

### 4.2 坐标变换：物理点保持不动，坐标表示发生变化

机器人学中的多坐标系建模涉及同一物理点在不同坐标系下的表示。

假设：

- $\{A\}$：基坐标系
- $\{B\}$：机器人末端坐标系
- $P$：空间中的某个物理点

点 $P$ 保持不动，其坐标表示随所采用的坐标系而变化。

记：

$$
P_B
$$

为点 $P$ 在坐标系 $\{B\}$ 下的坐标，

$$
P_A
$$

为同一个点 $P$ 在坐标系 $\{A\}$ 下的坐标。

如果 $T_{AB}$ 描述坐标系 $\{B\}$ 相对于坐标系 $\{A\}$ 的位姿，则：

$$
\boxed{P_A=T_{AB}P_B}
$$

展开后：

$$
P_A=R_{AB}P_B+t_{AB}
$$

其中：

- $R_{AB}$：坐标系 $\{B\}$ 相对于 $\{A\}$ 的朝向
- $t_{AB}$：坐标系 $\{B\}$ 原点在 $\{A\}$ 中的位置
- $P_B$：物理点 $P$ 在 $\{B\}$ 中的坐标表示
- $P_A$：同一个物理点 $P$ 在 $\{A\}$ 中的坐标表示

该变换改变点 $P$ 的坐标表示，物理位置保持不变：

$$
\boxed{\text{物理点保持不动，坐标表示随坐标系变化}}
$$

因此：

$$
P_B
\quad\xrightarrow{T_{AB}}\quad
P_A
$$

表示：

$$
\boxed{\text{点 }P\text{ 的 B 系坐标表示}
\rightarrow
\text{同一个点 }P\text{ 的 A 系坐标表示}}
$$

---

### 4.3 两种含义的对比

| 情况 | 坐标系 | 物理点 | 公式含义 |
|---|---|---|---|
| 主动变换 | 不变 | 实际运动 | $p' = Tp$ |
| 坐标变换 | 使用不同坐标系描述 | 保持不动 | $P_A=T_{AB}P_B$ |

齐次变换矩阵的物理含义取决于具体语境：

> **齐次变换矩阵提供旋转和平移的数学关系，可用于描述刚体实际运动，也可用于表示同一物理点在不同坐标系下的坐标转换；具体含义由物理语境确定。**

在机械臂运动学中，基坐标系、关节坐标系和末端坐标系之间的关系通常采用坐标变换表示：

$$
\boxed{\text{同一个物理对象不变，在不同坐标系之间转换其坐标表示}}
$$

#### 两种变换的关系

可概括为：

**主动变换：**

$$
\boxed{\text{坐标系保持不变，点发生运动}}
$$

**坐标变换：**

$$
\boxed{\text{物理点保持不动，切换坐标系表示}}
$$

二者的矩阵形式可能相似，但物理含义不同，应根据坐标系和物理对象的变化情况加以区分。

#### 三维具体写法

$$
\tilde{p} =
\begin{bmatrix}
x \\ y \\ z \\ 1
\end{bmatrix},
\quad
T =
\begin{bmatrix}
r_{11} & r_{12} & r_{13} & t_x \\
r_{21} & r_{22} & r_{23} & t_y \\
r_{31} & r_{32} & r_{33} & t_z \\
0 & 0 & 0 & 1
\end{bmatrix}
$$

则：

$$
\tilde{p}' = T \tilde{p}
=
\begin{bmatrix}
r_{11}x + r_{12}y + r_{13}z + t_x \\
r_{21}x + r_{22}y + r_{23}z + t_y \\
r_{31}x + r_{32}y + r_{33}z + t_z \\
1
\end{bmatrix}
$$

取前三个分量，就是变换后的三维点：

$$
p' =
\begin{bmatrix}
x' \\ y' \\ z'
\end{bmatrix}
=
R p + t
$$

#### 计算步骤

1. 将点表示为齐次坐标：

   $$
   p \rightarrow \tilde{p} = \begin{bmatrix} p \\ 1 \end{bmatrix}
   $$

2. 使用齐次变换矩阵左乘：

   $$
   \tilde{p}' = T \tilde{p}
   $$

3. 变换结果仍为齐次坐标：

   $$
   \tilde{p}' = \begin{bmatrix} p' \\ 1 \end{bmatrix}
   $$

4. 取前三个分量，得到普通三维点 $p'$。

5. 对列向量执行左乘，矩阵乘法顺序为：

   $$
   T \tilde{p} \neq \tilde{p} T
   $$

6. 总结：

   **齐次变换矩阵将旋转和平移统一为一次矩阵乘法；根据具体语境，它既可以描述刚体的实际运动，也可以描述同一物理对象在不同坐标系之间的坐标转换。**


```python
import numpy as np

def rot_x(angle):
    """生成绕X轴的旋转矩阵"""
    theta = np.radians(angle)
    c, s = np.cos(theta), np.sin(theta)
    return np.array([
        [1, 0,  0],
        [0, c, -s],
        [0, s,  c]
    ])

def rot_y(angle):
    """生成绕Y轴的旋转矩阵"""
    theta = np.radians(angle)
    c, s = np.cos(theta), np.sin(theta)
    return np.array([
        [ c, 0, s],
        [ 0, 1, 0],
        [-s, 0, c]
    ])

def rot_z(angle):
    """生成绕Z轴的旋转矩阵"""
    theta = np.radians(angle)
    c, s = np.cos(theta), np.sin(theta)
    return np.array([
        [c, -s, 0],
        [s,  c, 0],
        [0,  0, 1]
    ])

def get_homogeneous_matrix(R, P):
    """组合旋转和平移生成4x4齐次变换矩阵"""
    T = np.eye(4)
    # T[0:3, 0:3]，取0、1、2行和0、1、2列，赋值为旋转矩阵R
    T[:3, :3] = R
    # T[0:3, 3]，取0、1、2行和第3列，赋值为平移向量P
    T[:3, 3] = P
    return T
```

```python
# 场景参数示例
# 机器人末端坐标系 {B} 相对于基坐标系 {A} 的位姿：
# 绕 X 轴旋转 120 度
# 平移向量为 [3, 2, 1]，各分量单位为米

# 1. 定义旋转和平移
R_AB = rot_x(120)
P_AB = np.array([3, 2, 1])

# 2. 构建变换矩阵 T_AB
T_AB = get_homogeneous_matrix(R_AB, P_AB)

# 3. 定义末端上的一个点 P
P_B = np.array([1, 0, 0, 1]) # 齐次坐标

# 4. 计算该点在基坐标系中的位置
P_A = T_AB @ P_B

print("变换矩阵 T_AB:\n", T_AB)
print("-" * 30)
print(f"点在局部坐标系 B: {P_B[:3]}")
print(f"点在全局坐标系 A: {P_A[:3]}")
```
